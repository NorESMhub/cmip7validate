# Overview

Quick validation of cmorized NorESM3 data for CMIP7 for the following datasets:

````python
# load parameters
import yaml
from IPython.display import HTML, display
from pathlib import Path
from pyprojroot import here

# Get project root
root_path = Path(here())

with open(root_path.joinpath('params.yml'), 'r') as f:
    param = yaml.safe_load(f)
    cmorout = param['datasets']['cmorout']
    source_id = param['datasets']['source_id']
    experiment_id = param['datasets']['experiment_id']
    variant_label = param['datasets']['variant_label']
    version = param['datasets']['version']
    
print(f'\033[1mexperiment_id:\033[0m {experiment_id}')
print(f'\033[1msource_id:\033[0m {source_id}')
print(f'\033[1mvariant_label:\033[0m {variant_label}')
print(f'\033[1mdata_version:\033[0m {version}')
print(f'\033[1mdata_path:\033[0m {cmorout}/{source_id}/{experiment_id}/{variant_label}/{version}')
````
