### Experiment Information:

````python
# load parameters
import yaml
from IPython.display import HTML, display

with open('../params.yml', 'r') as f:
    param = yaml.safe_load(f)
    cmorout = param['cmorout']
    source_id = param['source_id']
    experiment_id = param['experiment_id']
    variant_label = param['variant_label']
    version = param['version']
    
print(f'\033[1mexperiment_id:\033[0m {experiment_id}')
print(f'\033[1msource_id:\033[0m {source_id}')
print(f'\033[1mvariant_label:\033[0m {variant_label}')
print(f'\033[1mdata_version:\033[0m {version}')
print(f'\033[1mdata_path:\033[0m {cmorout}/{source_id}/{experiment_id}/{variant_label}/{version}')
````
